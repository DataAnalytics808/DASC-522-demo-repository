##Principal components analysis & anomaly detection

DU

```python
random.seed(7)
x, _ = make_blobs(n_samples=200, centers=1, cluster_std=.3, center_box=(20, 5))

plt.scatter(x[:,0], x[:,1])
plt.show()
```

Model

```python
dbscan = DBSCAN(eps = 0.28, min_samples = 20)
print(dbscan)

pred = dbscan.fit_predict(x)
```

Analysis

```python
# pull out anomalies
anom_index = where(pred == -1)
values = x[anom_index]

plt.scatter(x[:,0], x[:,1])
plt.scatter(values[:,0], values[:,1], color='r')
plt.show()
```

Re-model & analyze with varying eps

```python
dbscan = DBSCAN(eps = 0.22, min_samples = 20)
print(dbscan)

pred = dbscan.fit_predict(x)

# pull out anomalies
anom_index = where(pred == -1)
values = x[anom_index]

plt.scatter(x[:,0], x[:,1])
plt.scatter(values[:,0], values[:,1], color='r')
plt.show()
```
