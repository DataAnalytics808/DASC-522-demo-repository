```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
sns.set_theme(style='darkgrid')  # set plotting style

df = pd.read_csv("https://raw.githubusercontent.com/DataAnalytics808/DASC-522-demo-repository/main/data/Z_X.csv")
dfy = pd.read_csv("https://raw.githubusercontent.com/DataAnalytics808/DASC-522-demo-repository/main/data/Z_y.csv")

#combine dataframes
df['THC'] = dfy['THC']


df2 = df[['THC', 'Age', 'Education', 'Nscore', 'Escore', 'Oscore', 'Ascore','Cscore', 'Impulsive', 'SS']]  # 10 variables in the dataframe
axes = df2.plot.hist(subplots=True,
                     layout=(5, 2),  # 5 x 2 grid of plots
                     figsize=(12,12),
                     legend=True,
                     cmap='viridis',  # colormap to look nice
                     fontsize=12)
```
