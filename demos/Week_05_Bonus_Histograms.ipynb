```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
```

## Bonus - Plotting histograms of numeric variables

### Load Wine Dataset


```python
# Load the wine dataset
wine_data = load_wine(as_frame=True)
df = wine_data.frame
df.head()
```

### Plot Histograms of Numeric Features


```python
sns.set_theme(style="darkgrid")

# Select 10 numeric features for a 5x2 grid
features = ["alcohol", "malic_acid", "ash", "alcalinity_of_ash", "magnesium",
            "total_phenols", "flavanoids", "color_intensity", "hue", "proline"]
df2 = df[features]
axes = df2.plot.hist(subplots=True,
                     layout=(5, 2),
                     figsize=(12, 12),
                     legend=True,
                     cmap="viridis",
                     fontsize=12)
plt.tight_layout()
plt.show()
```

### Alternative: Plotting Histograms for Drug Consumption Dataset (Z_X / Z_y)


```python
df_zx = pd.read_csv("https://raw.githubusercontent.com/DataAnalytics808/DASC-522-demo-repository/main/data/Z_X.csv")
df_zy = pd.read_csv("https://raw.githubusercontent.com/DataAnalytics808/DASC-522-demo-repository/main/data/Z_y.csv")

df_zx["THC"] = df_zy["THC"]
df_drugs = df_zx[["THC", "Age", "Education", "Nscore", "Escore", "Oscore", "Ascore", "Cscore", "Impulsive", "SS"]]
axes = df_drugs.plot.hist(subplots=True,
                          layout=(5, 2),
                          figsize=(12, 12),
                          legend=True,
                          cmap="viridis",
                          fontsize=12)
plt.tight_layout()
plt.show()
```
