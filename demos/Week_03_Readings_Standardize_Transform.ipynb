# Box-Cox Transformation
Feature transformation example using Scikit-Learn's `PowerTransformer` on the Wine dataset.

```python
from sklearn.datasets import load_wine
import pandas as pd
from sklearn.preprocessing import PowerTransformer
import matplotlib.pyplot as plt

wine = load_wine()
df = pd.DataFrame(data=wine.data, columns=wine.feature_names)
print(df.head())

df['malic_acid'].hist(bins=20)
plt.title("Original Malic Acid")
plt.show()

# Box-Cox transformation requires strictly positive values
pt = PowerTransformer(method="box-cox")
transformed = pt.fit_transform(df[['malic_acid']])
df['malic_acid_boxcox'] = transformed

df['malic_acid_boxcox'].hist(bins=20)
plt.title("Box-Cox Transformed Malic Acid")
plt.show()
```

# Standardize using Scikit-Learn
Standardizing features with `StandardScaler` to have zero mean and unit variance.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_wine
import pandas as pd

wine = load_wine()
X = wine.data
y = wine.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(X_scaled, columns=wine.feature_names)
print(df_scaled.head())
```

# Standardize using Keras
Incorporating feature normalization directly into a Keras model architecture via the `Normalization` preprocessing layer.

```python
import numpy as np
from sklearn.datasets import load_wine
from tensorflow.keras.layers import Normalization
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

wine = load_wine()
X = wine.data
y = wine.target

normalizer = Normalization(axis=-1)
normalizer.adapt(X)

model = Sequential([
    normalizer,
    Dense(10, activation='relu'),
    Dense(3, activation='softmax')
])

model.summary()
```
